# Statistical, Psychometric & Explainability Evaluation of Machine Learning & Multimodal Large Language Models (MLLM Experiments 1–4, Study 2)

This notebook evaluates the participant-level state anxiety predictions (`n = 92`, `specified_avrr92`) across both the **19-Feature Multimodal Machine Learning Benchmark** and the **Four-Experiment Progressive Modality Ablation in `mllm.ipynb`**:
- **Experiment 1 (`Video Only`)**: Silent video frames only; no audio or speech semantics.
- **Experiment 2 (`Video + Audio, Non-Semantic Vocal Acoustics Only`)**: Visual cues + non-semantic vocal acoustics (pitch/F0, vocal tone, jitter/shimmer, intensity, pauses, breathing); semantic meaning of spoken words strictly prohibited.
- **Experiment 3 (`Video + Audio + Speech Context`)**: Full audiovisual + spoken semantic context.
- **Experiment 4 (`Video + Audio + Speech Context + Physiological Indicators`)**: Augments Experiment 3 with concurrent Heart Rate / HRV (`mean_hr`, `rmssd`, `sdnn`, `sbp`, `dbp`) and Facial Thermal Imaging (`nose_tip_temperature`, `left_cheek_temperature`, `right_cheek_temperature`) via `HeartRateInterpreter` and `ThermalInterpreter`.

### Analytical Hierarchy
1. **Primary Focus — Predicting Pre-Test, Post-Test, and Subtracted Improvement (`Pre - Post`) State Anxiety Scores**: MAE, RMSE, Pearson $r$, Spearman $\rho$, and directional sign agreement against self-reported `STAI-AA` (Absence of Calmness & Security, range 10–40), `STAI-AP` (Presence of Acute Tension, range 10–40), and `STAI-Total` (range 20–80).
2. **Explainability of Machine Learning & MLLM (`736 Reasoning Traces`)**: Feature selection frequencies, standardized shrinkage weights, and quantitative cue-to-score attribution across visual, acoustic, semantic, and physiological channels.
3. **Downstream Theoretical Meaning of AI Predictions for Experimental Groups**: Evaluating whether blind-to-group ML and MLLM anxiety predictions act as an objective surrogate endpoint that independently reproduces the $2\times 2$ and Exposure Time-Dose (`10 min = 1.0, 5 min = 0.5`) experimental effects of Petting (`H1a`), Feeding (`H1b`), and Direct Interaction vs. Active Observation (`OBS`), and decoding modality-specific behavioral signatures.
